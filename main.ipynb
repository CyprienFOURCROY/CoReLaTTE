```python
import json

def variable_metadata_to_text(json_dict: dict) -> str:
    lines = []
    print("title json dict :", json_dict)

    for column_name, metadata in json_dict.items():
       
        meaning = metadata.get("meaning", "No meaning provided")
        if meaning == "No meaning provided":
            print(f"Warning: No meaning provided for column '{column_name}'.")
        values_label = metadata.get("values_label")

        lines.append(f'Column "{column_name}" means: {meaning}.')

        if values_label:
            lines.append("Possible values:")
            for label, value in values_label.items():
                lines.append(f"  - {value}: {label}")

        lines.append("")  # blank line between variables

    return "\n".join(lines)
```

```python
def get_required_tables(plan_json: dict) -> set[str]:
    plan = validate_query_plan(plan_json)

    required_tables = set()

    for node in plan.nodes:
        if node.operation == "scan":
            required_tables.add(node.table)

    return required_tables


def execute_query_plan(plan_json: dict, tables: dict):
    required_tables = get_required_tables(plan_json)

    missing_tables = required_tables - set(tables.keys())

    if missing_tables:
        raise ValueError(
            f"Missing required table(s): {sorted(missing_tables)}. "
            f"Available tables: {sorted(tables.keys())}"
        )

    code = compile_query_plan(plan_json)

    namespace = {}

    exec(code, namespace)

    run_query = namespace["run_query"]

    return run_query(tables)
```

```python
import json
from pathlib import Path

ROOT = Path.cwd()
JSON_DIR = ROOT / "raw_data" / "hh09dta_b2" / "codebook_json"


def variable_metadata_to_text(json_dict: dict) -> str:
    lines = []

    for column_name, metadata in json_dict.items():
        meaning = metadata.get("meaning", "No meaning provided")
        values_label = metadata.get("values_label")

        lines.append(f'Column "{column_name}" means: {meaning}.')

        if values_label:
            lines.append("Possible values:")
            for label, value in values_label.items():
                lines.append(f"  - {value}: {label}")

        lines.append("")

    return "\n".join(lines)


for json_path in JSON_DIR.glob("*_enriched.json"):
    txt_path = json_path.with_suffix(".txt")

    with open(json_path, "r", encoding="utf-8") as f:
        json_dict = json.load(f)

    text = variable_metadata_to_text(json_dict)

    txt_path.write_text(text, encoding="utf-8")

    print(f"Created: {txt_path.name}")
```
